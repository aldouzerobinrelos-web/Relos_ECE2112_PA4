# **C. CATEGORY-AVERAGE VISUALIZATION**

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(20, 4))
```

```python
board2['Average'] = (board2['Math'] + board2['Electronics'] + board2['GEAS'] + board2['Communication']) / 4
track_mean = board2.groupby('Track')['Average'].mean().reset_index()
gender_mean = board2.groupby('Gender')['Average'].mean().reset_index()
hometown_mean = board2.groupby('Hometown')['Average'].mean().reset_index()
```

**Track Average**

```python
display(track_mean)
```

**Gender Average**

```python
display(gender_mean)
```

**Hometown Average**

```python
display(hometown_mean)
```

**Bar Graphs for all 3**

```python
plt.figure(figsize=(20, 4))

plt.subplot(1, 3, 1)
plt.bar(board2['Track'], board2['Average'])
plt.title('Track')

plt.subplot(1, 3, 2)
plt.bar(board2['Gender'], board2['Average'])
plt.title('Gender')

plt.subplot(1, 3, 3)
plt.bar(board2['Hometown'], board2['Average'])
plt.title('Hometown')

plt.tight_layout()
plt.show()
```

The Microelectronics category has the highest sample mean Average for the track category.

The Female category has the highest sample mean Average for the gender category.

The Luzon category has the highest sample mean Average for the hometown category.
